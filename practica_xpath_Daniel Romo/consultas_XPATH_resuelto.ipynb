##### **Ejercicio 1. Utilización de XPath | XML**
En esta práctica se aprenderá a leer un documento XML utilizando la biblioteca **lxml** y a recuperar información específica mediante expresiones **XPath**.

Como primer paso se debe instalar la librería **lxml**

```python
######################################################################################################
# Bibliotecas
# lxml: Permite leer y consultar documentos XML mediante XPath.
######################################################################################################


# etree: Módulo de la biblioteca lxml utilizado para leer y procesar documentos XML.

from lxml import etree

# La función parse() carga el archivo XML en memoria y crea una estructura de árbol que permite navegar por sus elementos.

tree = etree.parse("pacientes.xml")

######################################################################################################
# XPATH | /pacientes/paciente/nombre/text()
# /pacientes - Selecciona el nodo raíz
# /paciente - Selecciona cada paciente
# /nombre - Selecciona el elemento nombre
# text() - Obtiene únicamente el contenido textual
######################################################################################################

nombres = tree.xpath(
    "/pacientes/paciente/nombre/text()"
)

#Imprime los valores encontrandos dentro del nodo nombre

print(nombres)
```

##### **Ejercicio 2. Utilización de XPath | XML**

Dentro de este ejercicio se utiliza una expresión XPath para localizar todos los nodos correspondientes a los pacientes registrados en el archivo pacientes.xml.

A través de un ciclo for, se recorre cada uno de los elementos <paciente> identificados por XPath y se extrae la información asociada a cada registro, incluyendo el identificador, nombre, edad, sexo y diagnóstico.

Para recuperar la información se emplean los métodos get() y findtext(), los cuales permiten acceder a los atributos y elementos hijos del documento XML. Finalmente, los datos obtenidos se muestran en pantalla, permitiendo visualizar de forma estructurada la información almacenada para cada paciente.

Este ejercicio tiene como propósito comprender la estructura jerárquica de un documento XML, aplicar consultas XPath y extraer información específica utilizando Python, sentando las bases para consultas más avanzadas y posteriores análisis de datos.

```python
######################################################################################################
# Bibliotecas
# lxml: Permite leer y consultar documentos XML mediante XPath.
######################################################################################################

from lxml import etree

# Variable tree - Parsea el archivo de tipo xml (pacientes.xml)

tree = etree.parse("pacientes.xml")

#######################################################################################################
# Se define el XPATH, se obtendrán todos los datos que se encuentre dentro de la estructura de paciente
# La instrucción get() permite acceder al valor de un atributo de un elemento XML.
# El método findtext() permite localizar un elemento hijo dentro de un nodo XML y recuperar directamente 
# su contenido textual.
########################################################################################################

for paciente in tree.xpath("/pacientes/paciente"):

    id_paciente = paciente.get("id")
    nombre = paciente.findtext("nombre")
    edad = paciente.findtext("edad")
    sexo = paciente.findtext("sexo")
    diagnostico = paciente.findtext("diagnostico")

    #Imprime los resultados encontados para cada uno de los pacientes

    print("ID:", id_paciente)
    print("Nombre:", nombre)
    print("Edad:", edad)
    print("Sexo:", sexo)
    print("Diagnóstico:", diagnostico)
    print("-"*30)
```

##### **Ejercicio 3. Utilización de XPath | XML**

El ejercicio tiene como objetivo filtrar únicamente aquellos pacientes cuyo diagnóstico sea **Diabetes**, permitiendo recuperar información específica sin recorrer manualmente todo el documento.

```python
######################################################################################################
# Bibliotecas
# lxml: Permite leer y consultar documentos XML mediante XPath.
######################################################################################################

from lxml import etree

# Variable tree - Parsea el archivo de tipo xml (pacientes.xml)

tree = etree.parse("pacientes.xml")

# XPATH
# /pacientes → Selecciona el nodo raíz.
# /paciente → Selecciona todos los pacientes.
# [diagnostico='Diabetes'] → Filtra únicamente los pacientes cuyo diagnóstico sea "Diabetes".
#
# Esto equivale en SQL:
#      SELECT *
#      FROM pacientes
#      WHERE diagnostico = 'Diabetes';

pacientes = tree.xpath(
    "/pacientes/paciente[diagnostico='Diabetes']"
)

# Extrae los pacientes con diagnóticos tipo diabetes 

for paciente in pacientes:

    print("ID:", paciente.get("id"))
    print("Nombre:", paciente.findtext("nombre"))
    print("Diagnóstico:", paciente.findtext("diagnostico"))
    print()
```

```python
from lxml import etree
import pandas as pd

# #####################################################
# Cargar y parsear el documento XML
# #####################################################

tree = etree.parse("pacientes.xml")

# #####################################################
# Consulta XPath básica
# Obtener todos los nodos paciente
# #####################################################

datos = []

for paciente in tree.xpath("/pacientes/paciente"):

    datos.append({
        "id": paciente.get("id"),
        "nombre": paciente.findtext("nombre"),
        "edad": paciente.findtext("edad"),
        "sexo": paciente.findtext("sexo"),
        "diagnostico": paciente.findtext("diagnostico")
    })

# #####################################################
# Crear DataFrame
# #####################################################

df = pd.DataFrame(datos)

print("=== DATAFRAME COMPLETO ===")
print(df)

# #####################################################
# Convertir la edad a valor numérico
# #####################################################

df["edad"] = pd.to_numeric(df["edad"])

# #####################################################
# CONSULTAS XPATH
# #####################################################

print("\n=== NOMBRES DE TODOS LOS PACIENTES ===")

nombres = tree.xpath("/pacientes/paciente/nombre/text()")

for nombre in nombres:
    print(nombre)

# #####################################################
# Obtener todos los IDs
# #####################################################

print("\n=== IDs DE LOS PACIENTES ===")

ids = tree.xpath("/pacientes/paciente/@id")

for identificador in ids:
    print(identificador)

# #####################################################
# Pacientes mayores de 40 años utilizando XPath
# #####################################################

print("\n=== PACIENTES MAYORES DE 40 AÑOS (XPath) ===")

pacientes_mayores = tree.xpath("/pacientes/paciente[edad > 40]")

for paciente in pacientes_mayores:

    print(
        paciente.findtext("nombre"),
        paciente.findtext("edad")
    )

# #####################################################
# Contar pacientes utilizando XPath
# #####################################################

print("\n=== TOTAL DE PACIENTES (XPath) ===")

total = tree.xpath("count(/pacientes/paciente)")

print(int(total))

# #####################################################
# Primer paciente
# #####################################################

print("\n=== PRIMER PACIENTE ===")

primer_paciente = tree.xpath("/pacientes/paciente[1]")

for p in primer_paciente:
    print(p.findtext("nombre"))

# #####################################################
# Último paciente
# #####################################################

print("\n=== ÚLTIMO PACIENTE ===")

ultimo_paciente = tree.xpath("/pacientes/paciente[last()]")

for p in ultimo_paciente:
    print(p.findtext("nombre"))


#######################################################
# Contar
#######################################################
total = tree.xpath("count(/pacientes/paciente)")

print("Total de pacientes:", int(total))


```

#### **Práctica XPATH | JSON**

La biblioteca universitaria almacena la información de sus libros en un documento XML denominado biblioteca.xml o biblioteca.json. Como analistas de datos, deberán utilizar expresiones XPath para consultar información contenida en dicho documento sin transformarla previamente a una estructura relacional.

A partir del archivo proporcionado, desarrollen las expresiones XPath necesarias para resolver los siguientes requerimientos

- Obtener todos los títulos
- Obtener todos los identificadores
- Obtener libros con precio mayor a 1000
- Obtener el último libro
- Contar libros


### **Solución de la práctica**

Para la solución se considera la siguiente estructura: el nodo raíz se llama `biblioteca`, contiene varios nodos `libro`, el identificador está guardado en el atributo `id` y cada libro posee los elementos `titulo` y `precio`.

Ejemplo de la estructura esperada: `<biblioteca><libro id="L001"><titulo>...</titulo><precio>1200</precio></libro></biblioteca>`.

> **Nota:** XPath se utiliza para consultar documentos XML. En la segunda parte se presenta la solución equivalente para JSON mediante las estructuras de Python.


```python
######################################################################################################
# SOLUCIÓN CON XPATH | XML
######################################################################################################

from lxml import etree

# Cargar y analizar el archivo XML.
tree = etree.parse("biblioteca.xml")

# 1. Obtener todos los títulos.
# text() devuelve únicamente el contenido textual de cada elemento titulo.
titulos = tree.xpath("/biblioteca/libro/titulo/text()")

print("=== TODOS LOS TÍTULOS ===")
for titulo in titulos:
    print(titulo)

# 2. Obtener todos los identificadores.
# @id selecciona el atributo id de cada elemento libro.
identificadores = tree.xpath("/biblioteca/libro/@id")

print("\n=== TODOS LOS IDENTIFICADORES ===")
for identificador in identificadores:
    print(identificador)

# 3. Obtener los libros cuyo precio sea mayor a 1000.
# [precio > 1000] es el predicado que aplica la condición.
libros_caros = tree.xpath("/biblioteca/libro[precio > 1000]")

print("\n=== LIBROS CON PRECIO MAYOR A 1000 ===")
for libro in libros_caros:
    print("ID:", libro.get("id"))
    print("Título:", libro.findtext("titulo"))
    print("Precio:", libro.findtext("precio"))
    print("-" * 30)

# 4. Obtener el último libro.
# last() devuelve la última posición del conjunto de nodos libro.
ultimo_libro = tree.xpath("/biblioteca/libro[last()]")

print("\n=== ÚLTIMO LIBRO ===")
if ultimo_libro:
    libro = ultimo_libro[0]
    print("ID:", libro.get("id"))
    print("Título:", libro.findtext("titulo"))
    print("Precio:", libro.findtext("precio"))
else:
    print("La biblioteca no contiene libros.")

# 5. Contar los libros.
# count() devuelve un número flotante; int() lo convierte en entero.
total_libros = int(tree.xpath("count(/biblioteca/libro)"))

print("\n=== TOTAL DE LIBROS ===")
print(total_libros)
```

### **Solución equivalente para JSON**

Se considera que `biblioteca.json` tiene un arreglo llamado `libros` y que cada objeto contiene las propiedades `id`, `titulo` y `precio`. En JSON no se utiliza XPath directamente; las consultas se realizan recorriendo la lista de libros.


```python
######################################################################################################
# SOLUCIÓN EQUIVALENTE | JSON
######################################################################################################

import json

# Cargar el archivo JSON.
with open("biblioteca.json", "r", encoding="utf-8") as archivo:
    datos_json = json.load(archivo)

libros = datos_json["biblioteca"]["libros"]

# 1. Obtener todos los títulos.
titulos = [libro["titulo"] for libro in libros]
print("=== TODOS LOS TÍTULOS ===")
for titulo in titulos:
    print(titulo)

# 2. Obtener todos los identificadores.
identificadores = [libro["id"] for libro in libros]
print("\n=== TODOS LOS IDENTIFICADORES ===")
for identificador in identificadores:
    print(identificador)

# 3. Obtener los libros cuyo precio sea mayor a 1000.
libros_caros = [libro for libro in libros if float(libro["precio"]) > 1000]
print("\n=== LIBROS CON PRECIO MAYOR A 1000 ===")
for libro in libros_caros:
    print("ID:", libro["id"])
    print("Título:", libro["titulo"])
    print("Precio:", libro["precio"])
    print("-" * 30)

# 4. Obtener el último libro.
print("\n=== ÚLTIMO LIBRO ===")
if libros:
    ultimo_libro = libros[-1]
    print("ID:", ultimo_libro["id"])
    print("Título:", ultimo_libro["titulo"])
    print("Precio:", ultimo_libro["precio"])
else:
    print("La biblioteca no contiene libros.")

# 5. Contar los libros.
total_libros = len(libros)
print("\n=== TOTAL DE LIBROS ===")
print(total_libros)
```
